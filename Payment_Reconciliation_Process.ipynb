```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
import random

np.random.seed(42)

random.seed(42)

num_transcations = 100

transcation_id = []

for i in range(1001, 1001 + num_transcations):
  txn_id = "TXN" + str(i).zfill(5)
  transcation_id.append(txn_id)
transcation_id[:5]
```

```python
start_date = datetime(2026,5,5)

transcation_amount = [199,299,349,499,599,699,799]

amounts = np.random.choice(transcation_amount, size=num_transcations)

print(f"Transcations amounts: {amounts}")
```

```python
timestamps = []

for i in range(num_transcations):
  rand_day = np.random.randint(0,2)
  rand_hours = np.random.randint(9,22)
  rand_minutes = np.random.randint(0,59)

  timestamp = start_date + timedelta(days = rand_day, hours = rand_hours, minutes = rand_minutes)

  timestamps.append(timestamp)

print(timestamps, end = ' ')
```

```python
method_options = ['UPI', 'CC', 'NetBaking', 'Wallet']
method_weights = [0.5, 0.3, 0.08, 0.12]

payments_methods = np.random.choice(method_options, size=num_transcations, p = method_weights)

payments_methods[:5]

```

```python
internal_ledger = pd.DataFrame({'transcation_id': transcation_id,
                                'amount':amounts, 'timestamp':timestamps,
                                'payment_methods':payments_methods,
                                'status': 'SUCCESS'})
```

```python
internal_ledger.head(5)
```

```python
gateway_export = internal_ledger.copy()

print(f"Start gateway export with {len(gateway_export)} transcations.")
```

```python
missing_indices = np.random.choice(gateway_export.index, size = 5, replace = False)

missing_txn_ids = []

for idx in missing_indices:
  txn_id = internal_ledger.loc[idx,'transcation_id']
  missing_txn_ids.append(txn_id)

print(f"Will remove these transcation ID's : {missing_txn_ids}")
```

```python
gateway_export = gateway_export.drop(missing_indices)
```

```python
mismatch_indices = np.random.choice(gateway_export.index, size = 3, replace = False)


fee_options = [10,20,50,100]

for idx in mismatch_indices:
  original_amount = gateway_export.loc[idx,'amount']
  fee_deducted = random.choice(fee_options)
  new_amount = original_amount - fee_deducted
  gateway_export.loc[idx, 'amount'] = new_amount

  txn_id = gateway_export.loc[idx, 'transcation_id']
  print(f"{txn_id}: ₹{original_amount} -> ₹{new_amount} (fee: ₹{fee_deducted})")
```

```python
extra_transcation_1 = {
    'transcation_id':'TXN99901',
    'amount' : 199,
    'timestamp' : start_date + timedelta(hours=14),
    'payment_method' : 'UPI',
    'status' : 'SUCCESS'

}

extra_transaction_2 = {
    'transaction_id': 'TXN99902',
    'amount': 2999,
    'timestamp': start_date + timedelta(days=1, hours=10),
    'payment_method': 'Card',
    'status': 'SUCCESS'
}

extra_transactions = pd.DataFrame(['extra_transcation_1','extra_transaction_2'])

gateway_export = pd.concat([gateway_export, extra_transactions], ignore_index = True)

print(f"Added extra transactions: TXN99901, TXN99902")
print(f"Gateway now has {len(gateway_export)} transactions")
```

```python
available_indices = gateway_export.index[:-2].tolist()
status_change_indices = np.random.choice(available_indices, size=2, replace=False)

for idx in status_change_indices:
    txn_id = gateway_export.loc[idx, 'transcation_id']
    gateway_export.loc[idx, 'status'] = 'FAILED'
    print(f"{txn_id}: Status changed to FAILED")
```

```python
print(f"\n{'='*50}")
print(f"Internal Ledger: {len(internal_ledger)} transactions")
print(f"Gateway Export: {len(gateway_export)} transactions")
print(f"{'='*50}")
```

```python
print("=== Internal Ledger Summary ===")
print(f"Date Range : {internal_ledger['timestamp'].min()} to {internal_ledger['timestamp'].max()}")
print(f"Total amount : {internal_ledger['amount'].sum():,}")
print(internal_ledger['payment_methods'].value_counts())
```

```python
print("\n === Gateway Export Summary === ")
print(f"Date range : {gateway_export['timestamp'].min()} to {gateway_export['timestamp'].max()}")
print(f"Total amount : {gateway_export['amount'].sum()}")
print(gateway_export['payment_methods'].value_counts())
```

```python
ledger_ids = set(internal_ledger['transcation_id'])
gateway_ids = set(gateway_export['transcation_id'])


print(f"Unique Ids in Ledger : {len(ledger_ids)}")
print(f"Unique Ids in Gateway : {len(gateway_ids)}")
```

```python
missing_in_gateway = ledger_ids - gateway_ids

print(f"\n Transcations in ledger but not in Gateway : {len(missing_in_gateway)}")
print(f"Transcation IDs : {missing_in_gateway}")
```

```python
missing_in_gateway_list = list(missing_in_gateway)

missing_in_gateway_df = internal_ledger[internal_ledger['transcation_id'].isin(missing_in_gateway_list)].copy()

missing_in_gateway_df['issue_type'] = 'MISSING_IN_GATEWAY'

total_missing_value = missing_in_gateway_df['amount'].sum()
print(f"\nTotal value at risk: ₹{total_missing_value:,}")
```

```python
missing_in_gateway_df
```

```python
common_ids = ledger_ids & gateway_ids

print(f"Transcations present in both : {len(common_ids)}")
```

```python
common_ids_list = list(common_ids)

ledger_common = internal_ledger[internal_ledger['transcation_id'].isin(common_ids_list)].copy()
gateway_common = gateway_export[gateway_export['transcation_id'].isin(common_ids_list)].copy()

print(f"Transcation common in ledger : {len(ledger_common)}")
print(f"Transcation common in gateway : {len(gateway_common)}")
```

```python
comparsion = ledger_common.merge(gateway_common[['transcation_id', 'amount',
                                                'status']], on= 'transcation_id',
                                 suffixes=('_ledger', '_gateway'))

print(f"Merged comparison table: {len(comparsion)} rows")
comparsion.head()
```

```python
comparsion['amount_diff'] = comparsion['amount_ledger'] - comparsion['amount_gateway']

amount_mismatches = comparsion[comparsion['amount_diff'] != 0].copy()

print(f"Transcations with amount mismatch : {len(amount_mismatches)}")

if len(amount_mismatches ) > 0:
  total_discrepancy = amount_mismatches['amount_diff'].sum()
  print(f"Total discrepancy : , Rs {total_discrepancy}")

  print("\n Mismatch Details : ")
  print(amount_mismatches[['transcation_id', 'amount_ledger', 'amount_gateway',
                           'amount_diff']])
```

```python
# Find status mismatches in our comparison table
status_mismatches = comparsion[comparsion['status_ledger'] != comparsion['status_gateway']].copy()

print(f"Transactions with status mismatch: {len(status_mismatches)}")

if len(status_mismatches) > 0:
    print("\n⚠️  CRITICAL — We think SUCCESS but gateway says otherwise:")
    print(status_mismatches[['transcation_id', 'amount_ledger', 'status_ledger', 'status_gateway']])
```

```python
amount_mismatch_report = amount_mismatches[['transcation_id', 'amount_ledger',
                                            'amount_gateway','amount_diff']].copy()

amount_mismatch_report['issue_type'] = 'AMOUNT_MISMATCH'
amount_mismatch_report.columns = ['transcation_id', 'ledger_amount',
                                  'gateway_amount', 'difference', 'issue_type']
print(f"Amount mismatch records : {len(amount_mismatch_report)}")

amount_mismatch_report
```
