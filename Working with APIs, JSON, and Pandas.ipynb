```python
# Simulated payment gateway response (like Razorpay or Stripe)
payment_data = {
    "payments": [
        {
            "payment_id": "pay_ABC123",
            "amount": 1500,
            "currency": "INR",
            "status": "captured",
            "method": "upi",
            "customer": {
                "id": "cust_001",
                "email": "priya@email.com",
                "contact": "9876543210"
            },
            "upi": {
                "vpa": "priya@okaxis"
            },
            "created_at": 1705320000
        },
        {
            "payment_id": "pay_DEF456",
            "amount": 3200,
            "currency": "INR",
            "status": "captured",
            "method": "card",
            "customer": {
                "id": "cust_002",
                "email": "rahul@email.com",
                "contact": "9876543211"
            },
            "card": {
                "last4": "4242",
                "network": "Visa"
            },
            "created_at": 1705406400
        }
    ],
    "count": 2,
    "has_more": False
}
```

```python
payments_list = payment_data["payments"]

df_payments = json_normalize(payments_list)

print(df_payments.head())
```

```python
print(df_payments.columns.tolist())
```

```python
# Step 3: Select and clean
df_payments_clean = df_payments[[ 'payment_id', 'amount', 'status', 'method','customer.email', 'created_at']].copy()

# Rename

df_payments_clean.columns = [ 'payment_id', 'amount', 'status', 'method','customer_email', 'timestamp']

df_payments_clean
```

```python
df_payments_clean['timestamp'] = pd.to_datetime(
                                   df_payments_clean['timestamp'], unit='s'
                                )

df_payments_clean
```

```python
# Get 30 days of Bitcoin price history

url_history = "https://api.coingecko.com/api/v3/coins/bitcoin/market_chart"

params = {
    "vs_currency": "inr",
    "days": "30",
    "interval": "daily"
}

response = requests.get(url_history, params=params)

history_data = response.json()

history_data
```
